```python
import tensorflow as tf 
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import pandas as pd 
import numpy as np 
```

```python
print(pd.__version__)
print(np.__version__)
print(tf.__version__)
```

```python
data=pd.read_csv('imdb_train.csv')

sample_data=10000
data=data.sample(sample_data,random_state=42).reset_index(drop=True)
```

```python
texts=data['review'].tolist()
tokenizer=Tokenizer()
tokenizer.fit_on_texts(texts)
```

```python
#tokenizer.word_index
```

```python
texts_small = texts[:1000]   # START SMALL
max_words = 30              # HARD LIMIT

input_seq = []

for sentence in texts_small:
    tokenized = tokenizer.texts_to_sequences([sentence])[0][:max_words]
    for i in range(1, len(tokenized)):
        input_seq.append(tokenized[:i+1])

max_len = max_words
```

```python
max_len=max([len(x) for x in input_seq])
padded_input_seq=pad_sequences(input_seq,maxlen=max_len,padding='pre')
```
